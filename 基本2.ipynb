```python
#find() rfind()
a='Hello, you are my favorite girl, can you be my girlfriend?'
print(a.find('girl'))
print(a.rfind('can'))
print(a.find('z'))
```

```python
#split() 分割字串
a='Hello, you! are my favorite girl, can you! be my girlfr!iend?'
print(a.split('!'))
print(a.split(' '))
```
