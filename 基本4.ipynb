````python
#好多＋問答式法
#好多個版
a=1+int(input('你想要數字到幾層(3~9,最多9層)?  ：'))
b=a-1
c=int(input('你想要幾個（橫向)?  ：'))
for i in range(1,a): #上半部
  print(c*(' '*(a-i-1)+str(i)*(2*i-1)+' '*(a-i-1)+' '*5))
  #print(' ')

for i in range(b-1,0,-1): #下半部
  print(c*(' '*(b-i)+str(i)*(2*i-1)+' '*(b-i)+' '*5))
  #print(' ')
````
