```python
a=['hehe','haha','huhu']
b=[1,2,3]
c=['hehe',1,2,['haha','huhu']]
d='hehe,haha'
al=list(d) #把每個字分開乘字串
bl=d.split(',') #以 ',' 分開乘字串
print(a)
print(b)
print(c)
print(al)
print(bl)

```

```python
a=[1,2,3,4,5,6,7,8,9]
print(a[5])
print(a[:5])
print(a[5:])
print(a[5:7])
print(a[-5:-1])
print(a[-1:-5]) #不行要print(a[-1::-1])
print(a[::2])
print(a[3::-1])
print(a[7::-1])
```
