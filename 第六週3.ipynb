```python
#練3
import math
a=int(input("三角形第一個邊長"))
b=int(input("三角形第二個邊長"))
c=int(input("三角形第三個邊長"))
d=(a+b+c)/2
if a+b>c and b+c>a and a+c>b:
  print(math.sqrt(d*(d-a)*(d-b)*(d-c)))
else:
  print("不構成三角形")
```
